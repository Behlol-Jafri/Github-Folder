```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
water_df = pd.read_csv('Watera.csv')
water_df.head()
```

```python
water_df.info()
```

```python
water_df.isna().sum()
```

```python
water_dataset = water_df.fillna(water_df.mean())
water_dataset.head()
```

```python
water_dataset.isna().sum()
```

```python
col_list = list(water_dataset.columns[:-1])
col_list
```

```python
x = water_dataset[col_list]
y = water_dataset['potability']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lg_model = LogisticRegression()
lg_model.fit(x_train, y_train)
```

```python
lg_pred = lg_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score , confusion_matrix

print('Accuracy', round(accuracy_score(y_test, lg_pred)*100,2))
```

```python
print(confusion_matrix(y_test, lg_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier()
dt_model.fit(x_train, y_train)
```

```python
dt_pred = dt_model.predict(x_test)
```

```python
print('Accuracy', round(accuracy_score(y_test, dt_pred)*100,2))
```

```python
print(confusion_matrix(y_test, dt_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)
```

```python
rf_pred = rf_model.predict(x_test)
```

```python
print('Accuracy', round(accuracy_score(y_test, rf_pred)*100,2))
```

```python
print(confusion_matrix(y_test, rf_pred))
```
